## HelaBERT Model Embedding Check

### Overview
This notebook demonstrates how to use the HelaBERT model to generate text embeddings for Sinhala language sentences. It loads a pre-trained BERT model, tokenizes Sinhala text using SentencePiece, and computes sentence embeddings using mean pooling aggregation.

### Key Components

#### 1. Tokenization
- **Tokenizer**: Uses SentencePiece unigram tokenizer (`unigram_32000_0.9995.model`)
- **Vocabulary Size**: 32,000 tokens
- **Purpose**: Converts raw Sinhala text into numerical token IDs that the model can process

#### 2. Pre-trained Model
- **Model**: HelaBERT_v02 (BERT-based model for Sinhala)
- **Architecture**: BERT without pooling layer
- **Hidden Dimension**: 384
- **Purpose**: Encodes tokenized text into contextual embeddings

#### 3. Embedding Generation
- **Method**: Mean pooling over all token embeddings
- **Output Shape**: [1, 384] for a single sentence
  - First dimension: batch size (1 sentence)
  - Second dimension: embedding dimension (384)

### Workflow

1. Initialize SentencePiece processor and load the tokenizer model
2. Load the pre-trained HelaBERT model in evaluation mode
3. Encode Sinhala text to token IDs
4. Pass tokens through the model to get token-level embeddings
5. Compute mean pooling to obtain a single sentence embedding
6. Display the shape of the resulting embedding

### Use Cases

- **Semantic Similarity**: Compare similarity between two Sinhala sentences
- **Text Classification**: Use embeddings as features for downstream tasks
- **Clustering**: Group similar Sinhala documents together
- **Information Retrieval**: Build semantic search systems for Sinhala text

```python
from transformers import BertModel
import torch
import sentencepiece as spm

TOKENIZER_MODEL = "../tokenizer/unigram_32000_0.9995.model"

sp = spm.SentencePieceProcessor()
sp.load(TOKENIZER_MODEL)
model = BertModel.from_pretrained("../helabert_v2/bert_final_model", add_pooling_layer=False)
model.eval()

text = "කෘත්‍රිම බුද්ධිය ශ්‍රී ලංකාවේ අනාගතය වෙනස් කරයි"
ids = torch.tensor([sp.encode(text, out_type=int)])

with torch.no_grad():
    out = model(ids)

# Mean pooling over all tokens
mean_embedding = out.last_hidden_state.mean(dim=1)  # [1, 384]
print(f"Mean pooling embedding shape: {mean_embedding.shape}")
```
